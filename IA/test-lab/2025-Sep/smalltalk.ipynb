```python
router_tester = [([('user', 'Hola'), ('ia', 'Hola que tal'), ('user', 'Todo bien y tu?')], 'small_talk'),
 ([('user', 'Estoy buscando un depa en boyacá')], 'extract'),
 ([('user', 'Quien es el presidente de Argentina')], 'other')]
```

```python
print("\n".join([role + ": " + message for role, message in [x for x in router_tester[0][0]]]))
```

```python
result_list = []

for test in router_tester:
    test_chat = "\n".join([role + ": " + message for role, message in [x for x in test[0]]])
    print(test_chat)
    result = chain.invoke(input={
        "message_context": test_chat
    })
    result_list.append(result)
```
